# ChromaDB (Vector DB) Demo

This notebook provides guidance on basic and advanced operations with ChromaDB for storing and searching financial news vectors.

## 1. Initialization

Set up Persistent Client and create a Collection.

```python
import chromadb
from chromadb.utils import embedding_functions

# Connect to ChromaDB Server running in Docker
client = chromadb.HttpClient(host="127.0.0.1", port=8000)

# Use the default embedding function (or can use SentenceTransformer)
default_ef = embedding_functions.DefaultEmbeddingFunction()

# Create or get the financial news collection
collection = client.get_or_create_collection(
    name="financial_news",
    embedding_function=default_ef
)

print("ChromaDB initialized and 'financial_news' collection created.")
```

## 2. Write Commands

Add news to the Vector DB with metadata.

```python
collection.add(
    documents=[
        "Bitcoin reaches a new all-time high above $70,000.",
        "Federal Reserve maintains interest rates at current levels.",
        "Nvidia stock surges after strong earnings report."
    ],
    metadatas=[
        {"symbol": "BTC", "sentiment": "positive", "source": "Reuters"},
        {"symbol": "ECON", "sentiment": "neutral", "source": "Bloomberg"},
        {"symbol": "NVDA", "sentiment": "positive", "source": "CNBC"}
    ],
    ids=["id1", "id2", "id3"]
)

print("Added 3 records to the collection.")
```

## 3. Query Commands

Search for news most similar to a query.

```python
results = collection.query(
    query_texts=["What is the latest news about crypto?"],
    n_results=2
)

print("Similar search results:")
for i, doc in enumerate(results['documents'][0]):
    print(f"- {doc} (Distance: {results['distances'][0][i]}) ")
```

## 4. Advanced

Search combined with Metadata filtering (e.g., get only BTC news).

```python
filtered_results = collection.query(
    query_texts=["market updates"],
    where={"symbol": "BTC"},
    n_results=1
)

print("Filtered results by Metadata (BTC only):")
print(filtered_results['documents'])
```

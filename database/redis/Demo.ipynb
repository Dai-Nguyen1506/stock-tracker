# Redis Database Demo

This notebook provides guidance on basic and advanced operations with Redis in the Stock Tracker project.

## 1. Initialization

Establish a connection to Redis.

```python
import redis
import json
import time

# Connect to Redis
r = redis.Redis(host='localhost', port=6379, db=0, decode_responses=True)

try:
    if r.ping():
        print("Successfully connected to Redis!")
except Exception as e:
    print(f"Connection error: {e}")
```

## 2. Write Commands

Operations with String, Hash and set TTL (Time To Live).

```python
# 1. Write simple string (String) with expiration time 60s
r.set('market_status', 'OPEN', ex=60)

# 2. Write complex object (Hash)
user_data = {
    'id': 1,
    'name': 'Antigravity',
    'last_login': time.time()
}
r.hset('user:100', mapping=user_data)

print("Sample data written to Redis.")
```

## 3. Query Commands

Read data from Redis.

```python
# Read String
status = r.get('market_status')
print(f"Market Status: {status}")

# Read Hash
user = r.hgetall('user:100')
print(f"User Data: {user}")
```

## 4. Advanced

Advanced techniques commonly used for Redis

```python
symbol = 'BTCUSDT'
zset_key = f'history:{symbol}'

# Add data to Sorted Set (Score is timestamp)
r.zadd(zset_key, {json.dumps({'price': 65000}): time.time()})
r.zadd(zset_key, {json.dumps({'price': 65100}): time.time() + 10})

# Get the latest 10 records
history = r.zrevrange(zset_key, 0, 9)
print(f"Latest price history for {symbol}:")
for item in history:
    print(json.loads(item))
```

Use distributed locks to ensure that at any given time only one process is allowed to synchronize data for a specific symbol
